```python
import numpy as np
```

##### creating array from scratch


```python
# creating a array from only zeros using zeros method

array_of_0=np.zeros((3,4))
print(array_of_0)
```

```python
# creating a array from only ones using ones method
array_of_1=np.ones((3,4))

print(array_of_1)
```

```python
# creating a array of random  numberss using random module
array_of_random_num=np.random.random((2,3))
print(array_of_random_num)
```

```python
# creating a array using range or sequnce numberss using arange method
array_of_sqeunce_num=np.arange(2,20,2)
print(array_of_sqeunce_num)
```
