```python
import numpy as np
```

##### creating array / vector  / matrix / and tensor

```python
#one dimensional array or a (   vector  )  
l1=[1,2,3,4,5,]

array_1=np.array(l1)
```

```python
# two dimensonal array or a  (  matrix  )
l2=[4,5,6]
l3=[7,8,9]

array_2=np.array([l2,l3])
```

```python
# creating three  dimensional array or a ( tensor )
l11=[1,2,3,4,5]
l22=[6,7,8,9,1]
l33=[9,8,2,1,2]

array_3=np.array([[l11,l22,l33]])
```
