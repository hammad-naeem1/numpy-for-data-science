##### array properties

```python
import numpy as np
```

```python
# example :
l1=[1,2,3]
l2=[4,5,6]
fd=np.array((l1,l2))
```

```python
                    # checking an array shape 
print(fd.shape)
# .shape is used to dertermine the shape on an array
```

```python
#                        reshaping an array 
new_array=fd.reshape(3,2)
print(new_array.shape)
#  reshape is used for reshaping an array 
```

```python
#                   creating a array using arange 
array_of_sqeunce_num=np.arange(12)
print(array_of_sqeunce_num)
array_of_sqeunce_num.reshape((3,4))
#  creating a array using range or sequnce numberss using arange method
```

```python
                # checking an array dimenstion 
print(fd.ndim)
# .ndim is used to show a array dimension
```

```python
#               checking total num of element in an array
print(fd.size)
# .size is used for  checking total num of element in an array
```

```python
#                checking a data type of an array
print(fd.dtype)
# .dtype is used for checking a data type of an array
```

```python
                    # flattening an array
# .flatten() creates a 1D copy of the array

flattened_array = array_of_sqeunce_num.flatten()
print(flattened_array)
# .flatten() is used to collapse a multi-dimensional array into 1D
```

```python
                    # flattening an array using ravel
# .ravel() returns a 1D view of the original array

raveled_array =array_of_sqeunce_num.ravel()
print(raveled_array)
# .ravel() is used to flatten an array without copying memory
```

```python
#                    reversing an array
# [::-1] or np.flip() reverses the sequence of elements
reversed_array = array_of_sqeunce_num[::-1]
# or: reversed_array = np.flip(fd)

print(reversed_array)
# [::-1] or np.flip() is used to reverse the order of elements
```
